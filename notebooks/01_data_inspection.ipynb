```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\Dell\OneDrive\Documents\Projects\sales_dashboard_superstore\data\raw\Sample_Superstore.csv",
    encoding='latin1',
    parse_dates=['Order Date', 'Ship Date']
)

print("Shape:", df.shape)
df.head(10)
```

```python
print(df.dtypes)
```

```python
# Check for missing values
print("Missing values per column:\n", df.isnull().sum())

# Check for duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())

# Quick overview of data types
print("Data types:\n", df.dtypes)

# Basic statistics for numeric columns
df.describe(include='all').T
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Convert numeric columns properly
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')

# Standardize string columns
df['State'] = df['State'].str.title().str.strip()
df['Category'] = df['Category'].str.title().str.strip()
df['Sub-Category'] = df['Sub-Category'].str.title().str.strip()
df['Region'] = df['Region'].str.title().str.strip()
```

```python
# Year, Month, YearMonth for time-series analysis
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['YearMonth'] = df['Order Date'].dt.to_period('M').astype(str)

# Profit Margin and processing time
df['Profit_Margin'] = df['Profit'] / df['Sales']
df['Processing_Days'] = (df['Ship Date'] - df['Order Date']).dt.days

# Sales per unit
df['Sales_per_Unit'] = df['Sales'] / df['Quantity']
```

```python
clean_path = r"C:\Users\Dell\OneDrive\Documents\Projects\sales_dashboard_superstore\data\clean\superstore_clean.csv"

df.to_csv(clean_path, index=False)
print(f"\n✅ Clean dataset saved successfully at:\n{clean_path}")
```
